# GPR model demo

```python
import gpflow
from gpflow.utilities import print_summary
# import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import tensorflow as tf

# The lines below are specific to the notebook format
# %matplotlib inline
# plt.rcParams["figure.figsize"] = (12, 6)
```

```python
N = 100  # number of time steps (scanning volumes)
D = 3  # number of time series (ROIs)
x = np.linspace(0, 1, N).reshape(-1, 1)
y = np.random.random(size=(N, D))
df = pd.DataFrame(y)

X = df.iloc[:, 0].values.reshape(-1, 1)
Y = df.iloc[:, 1].values.reshape(-1, 1)

# _ = plt.plot(X, Y, "kx", mew=2)
```

```python
k = gpflow.kernels.Matern52()
```

```python
print_summary(k)
```

```python
m = gpflow.models.GPR(
    data=(X, Y),
    kernel=k,
    mean_function=None,
)
```

```python
print_summary(m)
```

```python
m.likelihood.variance.assign(0.01)
m.kernel.lengthscales.assign(0.3)
```

```python
opt = gpflow.optimizers.Scipy()
```

```python
opt_logs = opt.minimize(
    m.training_loss, m.trainable_variables, options=dict(maxiter=100)
)
print_summary(m)
```
