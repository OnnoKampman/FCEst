# Static functional connectivity (sFC) estimation demo

Simple demo on how to estimate static functional connectivity using FCEst.

## Setup & Imports

Install the `FCEst` package using `pip install -e .`.

```python
%load_ext autoreload
%autoreload 2

import sys
```

```python
import matplotlib
import numpy as np
```

```python
np.random.seed(42)  # for reproducibility
```

```python
from fcest.models.sliding_windows import SlidingWindows
```

```python
sys.path.append('../../FCEst-benchmarking')

from helpers.synthetic_covariance_structures import get_periodic_covariances
from helpers.synthetic_covariance_structures import get_d2_covariance_structure
from helpers.simulations import simulate_time_series
```

```python
%matplotlib inline
matplotlib.rcParams['figure.figsize'] = (12, 6)
plt = matplotlib.pyplot
```

## Generate bivariate data

```python
N = 400
```

```python
cov_structure = get_d2_covariance_structure(
    get_periodic_covariances(num_samples=N, num_periods=1)
)
cov_structure.shape
```

```python
y = simulate_time_series(cov_structure)
y.shape
```

```python
xx = np.linspace(0, 1, N).reshape(-1, 1)
xx.shape
```

## Estimate sFC

```python
sw = SlidingWindows(
    x_train_locations=xx,
    y_train_locations=y,
)
estimated_tvfc_array = sw.estimate_static_functional_connectivity(
    connectivity_metric='correlation'
)  # (N, D, D)

# Select a single covariance term.
estimated_cov_array = estimated_tvfc_array[:, 0, 1]  # (N, )
```

## Plot estimates

```python
plt.plot(
    xx, 
    estimated_cov_array,
    # color=plot_color,
    # linewidth=linewidth,
    # label=label,
    # alpha=plot_alpha,
)
```
